```python
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)

IQR = Q3 - Q1

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
```

```python
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Limite inférieure :", lower)
print("Limite supérieure :", upper)
```

```python
outliers = df[(df['Age'] < lower) | (df['Age'] > upper)]

print("Nombre des Outliers :", len(outliers))
```

```python
sns.countplot(x='Survived', data=df)
plt.title('Distribution des classes')
plt.show()
```

```python
from sklearn.utils import resample

df_majority = df[df['Survived'] == 0]
df_minority = df[df['Survived'] == 1]

df_majority_downsampled = resample(
    df_majority,
    replace=False,
    n_samples=len(df_minority),
    random_state=42
)

df_balanced = pd.concat([df_majority_downsampled, df_minority])
```

```python
df_balanced['Survived'].value_counts()
```

```python
sns.countplot(x='Survived', data=df_balanced)
plt.title('Distribution après Undersampling')
plt.show()
```

```python
from imblearn.over_sampling import SMOTE

X = df[['Pclass', 'Age', 'SibSp', 'Parch', 'Fare',
        'Sex_female', 'Sex_male',
        'Embarked_C', 'Embarked_Q', 'Embarked_S']]

y = df['Survived']
```

```python
smote = SMOTE(random_state=42)

X_resampled, y_resampled = smote.fit_resample(X, y)
```

```python
print("Avant SMOTE :")
print(y.value_counts())

print("\nAprès SMOTE :")
print(y_resampled.value_counts())
```
